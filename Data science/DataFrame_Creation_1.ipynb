## Creating DataFrames in pandas is one of the fundamental tasks in data analysis. Below are different ways to create a DataFrame in pandas with examples.

    1. Creating a DataFrame from a Dictionary

```python
import pandas as pd
df=pd.DataFrame(columns=['name','age'])
print(df)
```

```python
# Creating a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
data
```

```python
# Creating a DataFrame
df = pd.DataFrame(data)

# Displaying the DataFrame
df
```

    2. Creating a DataFrame from a List of Lists

```python
# List of lists
data = [['Alice', 25, 'New York'],
        ['Bob', 30, 'Los Angeles'],
        ['Charlie', 35, 'Chicago']]

# Creating a DataFrame with column names
df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])

print(df)
```

    3. Creating a DataFrame from a List of Dictionaries

```python
# List of dictionaries
data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
    {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'}
]

# Creating a DataFrame
df = pd.DataFrame(data)

print(df)
```

    4. Creating a DataFrame from a NumPy Array

```python
import numpy as np

# Creating a NumPy array
array_data = np.array([
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago']
])

# Creating a DataFrame
df = pd.DataFrame(array_data, columns=['Name', 'Age', 'City'])

print(df)
```

    5. Creating an Empty DataFrame and Adding Data Later

```python
# Creating an empty DataFrame
df = pd.DataFrame(columns=['Name', 'Age', 'City'])

# Adding data using loc
df.loc[0] = ['Alice', 25, 'New York']
df.loc[1] = ['Bob', 30, 'Los Angeles']
df.loc[2] = ['Charlie', 35, 'Chicago']

print(df)
```

    6. Creating a DataFrame with a Custom Index

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data, index=['A', 'B', 'C'])

print(df)
```

    7. Creating a DataFrame from a Dictionary of Series

```python
data = {
    'Name': pd.Series(['Alice', 'Bob', 'Charlie']),
    'Age': pd.Series([25, 30, 35]),
    'City': pd.Series(['New York', 'Los Angeles', 'Chicago'])
}

df = pd.DataFrame(data)

print(df)
```
