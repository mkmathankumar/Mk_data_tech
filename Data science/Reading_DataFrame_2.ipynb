If you have data stored in a JSON file, you can read it using **pd.read_json():**

    (a) Reading JSON from a File

    (b) Creating DataFrame from JSON String

```python
import json

# JSON string
json_data = '''
[
    {"Name": "Alice", "Age": 25, "City": "New York"},
    {"Name": "Bob", "Age": 30, "City": "Los Angeles"},
    {"Name": "Charlie", "Age": 35, "City": "Chicago"}
]
'''

# Converting JSON string to DataFrame
df = pd.DataFrame(json.loads(json_data))

print(df)
```
